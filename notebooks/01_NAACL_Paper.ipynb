```python
mt_model_names = [
    "Phi-3.5-mini-instruct",
    "Mistral-7B-v0.3-Chinese-Chat",
    "Llama3.1-70B-Chinese-Chat",
]

model_names = []
for key in webqsp_result.keys():
    model_name = key.split("(")[0]
    if model_name not in model_names:
        model_names.append(model_name)
        # print(f"model_name: {shorter(model_name)}")
model_names = model_names + mt_model_names


def shorter(model_name):
    return (
        model_name.replace("Meta-", "")
        .replace("-Instruct", "")
        .replace("-instruct", "")
        .replace("-chat-hf", "")
        .replace("-Chinese-Chat", "")
        .replace("-it", "")
        .replace("gemma", "Gemma")
        .replace("-128k", "")
        .replace("b", "B")
    )


disply_model_names = [shorter(model_name) for model_name in model_names]
for m, d in zip(model_names, disply_model_names):
    print(f"{m} -> {d}")
```

```python
def detect_repetition_scores(row, col, debug=False):
    text = row[col] if isinstance(row[col], str) else ""

    newline_score, repetition_score, total_repetitions = detect_scores(
        row, debug=debug, answer_col=col, ground_truth_col="english"
    )

    return pd.Series(
        [
            newline_score if newline_score > 0 else 0,
            repetition_score if repetition_score > 0 else 0,
            total_repetitions if total_repetitions > 0 else 0,
            len(text),
        ]
    )


def load_mac_result(csv_file):
    df = pd.read_csv(csv_file)
    df2 = df.copy()
    # print(df2["model"].unique())
    df2["model"] = df2["model"].apply(lambda x: x.split("/")[-1])
    # print(df2["model"].unique())
    df2 = df2[df2["model"].isin(mt_model_names)]
    df2["repetition_penalty"] = df2["rpp"]
    if not df2.equals(df):
        df2.to_csv(csv_file, index=False)
        print(f"saved {csv_file}")
    return df2
```

```python
mappings = {
    "total_repetitions": "MTR",
    "rpp": "Repetition Penalty Parameter",
    "newline_score": "Mean NWC Repetitions",
    "repetition_score": "Mean Text Repetitions",
    "answer_len": "Mean Answer Length",
    "nrr": "NRR",  # "Non-Repetitive Ratio",
    "rr": "RR (in %)",  # "Repetitive Ratio",
    "comet": "COMET",
}


def find_display_name(column, dataset_name, rap_only=True):
    for k, v in mappings.items():
        if k in column:
            return v
    ret = (
        "F1"
        if "WebQSP" in dataset_name
        else "BERT-F1" if "MS MARCO" in dataset_name else "COMET"
    )
    return ret if "perf" in column else f"RAP-{ret}" if rap_only else f"{ret} and RAP"
```

```python
# plot the results
import matplotlib.pyplot as plt
import seaborn as sns
import re


def plot_detailed_repetitions(results, model_name=None, title=True):
    # make the font size bigger
    sns.set_context("poster")

    markers = ["D", "o", "s", "x", "P", "v", "X", "p", "*", "h"]
    colors = [
        "blue",
        "orange",
        "green",
        "red",
        "purple",
        "brown",
        "pink",
        "gray",
        "olive",
        "cyan",
    ]
    bar_colors = ["blue", "orange"]

    fig, ax = plt.subplots(3, 3, figsize=(30, 10))
    # print(f"ax: {ax.shape}")
    columns = [
        "newline_score",
        "repetition_score",
        # "answer_len",
    ]  #

    model_name_pattern = re.compile(model_name)

    for row, dataset_name in enumerate(results.keys()):
        for col, model_name_ext in enumerate(model_name_exts):
            for key in results[dataset_name].keys():
                if not model_name_pattern.search(key) or model_name_ext not in key:
                    continue

                print(f"key: {key}, model_name_ext: {model_name_ext}")

                df = results[dataset_name][key]["df_overall"]
                # print("newline_score len:", len(df["newline_score"]))

                col = col % 3

                current_ax = ax[row][col]

                bottom = 0
                for i, column in enumerate(columns):
                    # print(
                    #     f"key: {key}, model_name_ext: {model_name_ext}, column: {column}"
                    # )
                    bar1 = sns.barplot(
                        x="repetition_penalty",
                        y=column,
                        data=df,
                        color=colors[i],
                        ax=current_ax,
                        bottom=bottom,
                        label=find_display_name(column, dataset_name),
                        legend=False,
                    )
                    bottom += df[column]

                xticklabels = [
                    f"{1 + label * 0.02:.2f}" for label in current_ax.get_xticks()
                ]
                current_ax.set_xticklabels(xticklabels, rotation=90)

                xlabel = "Repetition Penalty Parameter" if row == 1 else None
                ax[row][col].set_xlabel(xlabel)

                ylabel = "Mean Repetition" if col == 0 else None
                ax[row][col].set_ylabel(ylabel)

                ax[row][col].set_title(
                    f"({chr(97 + row * 3 + col)}) {dataset_name} - {model_name_ext}"
                )
    if title:
        fig.suptitle(f"Model - {shorter(model_name)}")

    ax[2][2].set_visible(False)

    handles, labels = ax[row][col].get_legend_handles_labels()
    fig.legend(
        handles,
        labels,
        bbox_to_anchor=(0.7, 0.2),
        loc="center left",
    )

    # make the plot look nice
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D


def plot_all_models(results):
    # make the font size bigger
    sns.set_context("poster")  # Adjusts overall font size for large plots

    # 12 markers and 12 colors
    markers = ["s", "o", "D", "^", "v", "<", ">", "p", "*", "h", "X", "P"]
    colors = [
        "#0072B2",
        "#D55E00",
        "#009E73",
        "#F0E442",
        "#56B4E9",
        "#E69F00",
        "#CC79A7",
        "#000000",
        "#999999",
        "#F781BF",
        "#A6761D",
        "#1B9E77",
    ]

    fig, ax = plt.subplots(6, 3, figsize=(30, 40))
    columns = ["rrp", "perf"] * 3
    subplot_model_names = [model_names[:9]] * 4 + [mt_model_names] * 2
    dataset_names = list(results.keys())

    for row, column in enumerate(columns):
        for col, model_name_ext in enumerate(model_name_exts):
            col = col % 3
            for index, model_name in enumerate(subplot_model_names[row]):
                dataset_name = dataset_names[row // 2]
                key = f"{model_name}({model_name_ext})"
                if key not in results[dataset_name]:
                    continue
                df = results[dataset_name][key]["df_overall"]
                sns.lineplot(
                    x="repetition_penalty",
                    y=column,
                    data=df,
                    ax=ax[row][col],
                    label=shorter(model_name),
                    marker=markers[index if row < 4 else index + 9],
                    color=colors[index if row < 4 else index + 9],
                    legend=False,
                )

                fontsize = 30  # Increased fontsize
                # Set x and y labels with increased fontsize
                xlabel = (
                    "Repetition Penalty Parameter" if row == len(columns) - 1 else None
                )
                ax[row][col].set_xlabel(xlabel, fontsize=fontsize)  # Increased fontsize

                ylabel = find_display_name(column, dataset_name)
                ax[row][col].set_ylabel(ylabel, fontsize=fontsize)  # Increased fontsize

                # Set title with increased fontsize
                ax[row][col].set_title(
                    f"({chr(97 + row * 3 + col)}) {dataset_name} - {model_name_ext}",
                    fontsize=fontsize,  # Increased fontsize for titles
                )

    ax[4][2].set_visible(False)
    ax[5][2].set_visible(False)

    # Add horizontal lines between rows 2 & 3, and 4 & 5
    # Draw lines using Line2D and manually set the y-coordinates
    fig.add_artist(
        Line2D(
            [0, 1], [0.67, 0.67], color="black", linewidth=2, transform=fig.transFigure
        )
    )  # Line between rows 2 & 3
    fig.add_artist(
        Line2D(
            [0, 1],
            [0.343, 0.343],
            color="black",
            linewidth=2,
            transform=fig.transFigure,
        )
    )  # Line between rows 4 & 5

    # Adjust legend font size
    handles, labels = ax[row - 2][col].get_legend_handles_labels()
    handles2, labels2 = ax[row][col].get_legend_handles_labels()
    fig.legend(
        handles + handles2,
        labels + labels2,
        bbox_to_anchor=(0.75, 0.17),
        loc="center left",
        prop={"size": 40},
    )

    # make the plot look nice
    plt.tight_layout()
    plt.show()
```

```python
# plot the results
import matplotlib.pyplot as plt
import seaborn as sns


def plot_rpp_tuning(results, model_name=None, title=True, repetition_metric="rrp"):
    # make the font size bigger
    sns.set_context("poster")

    markers = ["D", "o", "s", "x", "P", "v", "X", "p", "*", "h"]
    colors = [
        "blue",
        "orange",
        "green",
        "red",
        "purple",
        "brown",
        "pink",
        "gray",
        "olive",
        "cyan",
    ]
    bar_colors = ["blue", "orange"]

    fig, ax = plt.subplots(3, 3, figsize=(30, 15))
    # print(f"ax: {ax.shape}")
    columns = ["perf", "rap"]

    fontsize = 30  # Increased fontsize

    for row, dataset_name in enumerate(results.keys()):
        for col, model_name_ext in enumerate(model_name_exts):
            col = col % 3
            for key in results[dataset_name].keys():
                if model_name not in key or model_name_ext not in key:
                    continue

                df = results[dataset_name][key]["df_overall"]
                # print(df.head())

                best_perf_indices = []

                for i, column in enumerate(columns):
                    # print(f"i: {i}, model_name_ext: {model_name_ext}, column: {column}")
                    sns.lineplot(
                        x="repetition_penalty",
                        y=column,
                        data=df,
                        ax=ax[row][col],
                        label=find_display_name(column, dataset_name),
                        marker=markers[i],
                        color=colors[i],
                        legend=False,
                    )

                    best_perf_index = df[column].idxmax()
                    best_perf_indices.append(best_perf_index)
                    delta = 0.01 if row < 2 else 0.01 / 3
                    ax[row][col].axvspan(
                        df["repetition_penalty"][best_perf_index] - delta,
                        df["repetition_penalty"][best_perf_index] + delta,
                        alpha=0.5,
                        edgecolor="none",
                        facecolor=bar_colors[i],
                    )

                twin = ax[row][col].twinx()
                repetition_metric_title = find_display_name(
                    repetition_metric, dataset_name
                )
                sns.lineplot(
                    x="repetition_penalty",
                    y=repetition_metric,
                    data=df,
                    ax=twin,
                    label=repetition_metric_title,
                    marker=markers[2],
                    color=colors[2],
                    legend=False,
                )

                twin.set_ylabel(repetition_metric_title)

                # add values on lineplot
                for i, p in enumerate(ax[row][col].lines):
                    best_perf_index = best_perf_indices[i]
                    for j, column in enumerate(columns):
                        for rpp in p.get_xdata():
                            if rpp == df["repetition_penalty"][best_perf_index]:
                                print(f"rpp: {rpp:.2f}")
                                val = df["perf"][best_perf_index]
                                print(f"perf: {val:.3f}")
                                ax[row][col].annotate(
                                    f"{val:.2f}",
                                    (rpp, val),
                                    ha="center",
                                    va="center",
                                    xytext=(0, 15),
                                    textcoords="offset points",
                                    color=colors[0],
                                )
                                val = df["rap"][best_perf_index]
                                print(f"rap: {val:.3f}")
                                ax[row][col].annotate(
                                    f"{val:.2f}",
                                    (rpp, val),
                                    ha="center",
                                    va="center",
                                    xytext=(0, -20),
                                    textcoords="offset points",
                                    color=colors[1],
                                )

                for _, p in enumerate(twin.lines):
                    for i, column in enumerate(columns):
                        best_perf_index = best_perf_indices[i]
                        for rpp in p.get_xdata():
                            if rpp == df["repetition_penalty"][best_perf_index]:
                                print(f"rpp: {rpp:.2f}")
                                val = df[repetition_metric][best_perf_index]
                                print(f"rr: {val:.2f}%")
                                twin.annotate(
                                    (
                                        f"{val:.2f}"
                                        if repetition_metric != "total_repetitions"
                                        else f"{val:.1f}"
                                    ),
                                    (rpp, val),
                                    ha="center",
                                    va="center",
                                    xytext=(0, 15),
                                    textcoords="offset points",
                                    color=colors[2],
                                )
                # increase y-axis limits
                y_limits = ax[row][col].get_ylim()
                ax[row][col].set_ylim(
                    [
                        y_limits[0] - (y_limits[1] - y_limits[0]) * 0.1,
                        y_limits[1] + (y_limits[1] - y_limits[0]) * 0.1,
                    ]
                )

                y_limits = twin.get_ylim()
                twin.set_ylim(
                    [
                        y_limits[0] - (y_limits[1] - y_limits[0]) * 0.1,
                        y_limits[1]
                        + (y_limits[1] - y_limits[0])
                        * (0.3 if row in [1, 2] and col == 1 else 0.1),
                    ]
                )

                # Set x and y labels with increased fontsize
                xlabel = "Repetition Penalty Parameter" if row == 2 else None
                ax[row][col].set_xlabel(xlabel, fontsize=fontsize)  # Increased fontsize

                ylabel = find_display_name(column, dataset_name, rap_only=False)
                ax[row][col].set_ylabel(ylabel, fontsize=fontsize)

                sub_title = (
                    f"({chr(97 + row * 3 + col)}) {dataset_name} - {model_name_ext}"
                )

                print(f"*** sub_title: {sub_title}")

                ax[row][col].set_title(sub_title, fontsize=fontsize)

    ax[2][2].set_visible(False)

    if title:
        fig.suptitle(f"Model - {shorter(model_name)}", fontsize=fontsize)

    handles, labels = ax[0][0].get_legend_handles_labels()
    handles2, labels2 = twin.get_legend_handles_labels()
    labels[0] = "Original Performance"
    labels[1] = "RAP"
    fig.legend(
        handles + handles2,
        labels + labels2,
        bbox_to_anchor=(0.7, 0.17),
        loc="center left",
        prop={"size": 40},
    )

    # make the plot look nice
    plt.tight_layout()
    plt.show()
```
